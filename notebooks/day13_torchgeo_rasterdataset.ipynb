```python
from torchgeo.datasets.utils import BoundingBox
```

```python
bbox = BoundingBox(
    minx=0,
    maxx=100,
    miny=0,
    maxy=100,
    mint=0,
    maxt=1
)

print(bbox)
```

```python
print("minx:", bbox.minx)
print("maxx:", bbox.maxx)

print("miny:", bbox.miny)
print("maxy:", bbox.maxy)

print("mint:", bbox.mint)
print("maxt:", bbox.maxt)
```
