```python
from torchgeo.samplers import (
    RandomGeoSampler,
    GridGeoSampler
)

print("GeoSamplers Imported Successfully")
```

```python
print("""
Day 14 Goal

Learn how TorchGeo converts
large geospatial rasters into
training patches.

Training:
RandomGeoSampler

Inference:
GridGeoSampler
""")
```

```python
scene_width = 10980
scene_height = 10980

print("Scene Width :", scene_width)
print("Scene Height:", scene_height)
```

```python
patch_size = 256

print(
    f"Patch Size = {patch_size} x {patch_size}"
)
```

```python
scene_area = (
    scene_width
    *
    scene_height
)

patch_area = (
    patch_size
    *
    patch_size
)

print("Scene Area :", scene_area)
print("Patch Area :", patch_area)
```

```python
approx_patches = (
    scene_width // patch_size
) * (
    scene_height // patch_size
)

print(
    "Approx Grid Patches:",
    approx_patches
)
```

```python
print("""
RandomGeoSampler

Patch Locations

Epoch 1
---------
Patch A
Patch B
Patch C

Epoch 2
---------
Patch D
Patch E
Patch F

Different every epoch.
""")
```

```python
print("""
GridGeoSampler

+----+----+----+
| A  | B  | C  |
+----+----+----+
| D  | E  | F  |
+----+----+----+

Covers entire scene.
""")
```

```python
print("""
GeoSentinel Training

Sentinel Image
        ↓
RandomGeoSampler
        ↓
Training Patches
        ↓
Model Training

GeoSentinel Inference

Sentinel Image
        ↓
GridGeoSampler
        ↓
Prediction Patches
        ↓
Urban Expansion Map
""")
```

```python
sizes = [128, 256, 512]

for size in sizes:

    patches = (
        scene_width // size
    ) * (
        scene_height // size
    )

    print(
        f"Patch Size {size}: "
        f"{patches} patches"
    )
```

```python
print("""
Key Learning

GeoSamplers do not sample
image indices.

GeoSamplers sample
geographic regions.

This is what makes TorchGeo
different from standard PyTorch.
""")
```
