```python
from torch.utils.data import Dataset
from torch.utils.data import DataLoader

import torch
```

```python
class DummyGeoDataset(Dataset):

    def __init__(self, size=100):

        self.size = size

    def __len__(self):

        return self.size

    def __getitem__(self, idx):

        image = torch.rand(
            4,
            256,
            256
        )

        mask = torch.randint(
            0,
            2,
            (256,256)
        )

        return {
            "image": image,
            "mask": mask
        }
```

```python
dataset = DummyGeoDataset()

print(
    "Dataset Size:",
    len(dataset)
)
```

```python
sample = dataset[0]

print(sample.keys())

print(
    sample["image"].shape
)

print(
    sample["mask"].shape
)
```

```python
loader = DataLoader(
    dataset,
    batch_size=4,
    shuffle=True
)
```

```python
batch = next(
    iter(loader)
)

print(
    batch["image"].shape
)

print(
    batch["mask"].shape
)
```

```python
print("""
Batch Shape

(B,C,H,W)

B = Batch Size

C = Channels

H = Height

W = Width
""")
```

```python
batch_sizes = [
    1,
    2,
    4,
    8
]

for bs in batch_sizes:

    loader = DataLoader(
        dataset,
        batch_size=bs
    )

    batch = next(
        iter(loader)
    )

    print(
        f"Batch Size {bs}:",
        batch["image"].shape
    )
```
