```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = np.random.rand(
    256,
    256,
    3
)

plt.figure(figsize=(6,6))

plt.imshow(image)

plt.title(
    "Synthetic Satellite Image"
)

plt.axis("off")

plt.show()
```

```python
mask = np.zeros(
    (256,256),
    dtype=np.uint8
)

mask[50:120, 50:120] = 1

mask[130:200, 100:220] = 2

mask[30:80, 170:230] = 3
```

```python
plt.figure(figsize=(6,6))

plt.imshow(mask)

plt.title(
    "Segmentation Mask"
)

plt.colorbar()

plt.show()
```

```python
fig, ax = plt.subplots(
    1,
    2,
    figsize=(12,6)
)

ax[0].imshow(image)
ax[0].set_title(
    "Image"
)

ax[1].imshow(mask)
ax[1].set_title(
    "Mask"
)

plt.show()
```

```python
classes, counts = np.unique(
    mask,
    return_counts=True
)

for cls, count in zip(
    classes,
    counts
):
    print(
        f"Class {cls}: {count}"
    )
```

```python
class_names = {
    0:"Background",
    1:"Vegetation",
    2:"Urban",
    3:"Water"
}

for cls in classes:

    print(
        cls,
        class_names[cls]
    )
```

```python
print("""
GeoSentinel Classes

0 = Background

1 = Vegetation

2 = Urban

3 = Water

Future:

4 = Bare Soil

5 = Road
""")
```

```python
prediction = mask.copy()

prediction[
    40:60,
    40:60
] = 2
```

```python
fig, ax = plt.subplots(
    1,
    2,
    figsize=(12,6)
)

ax[0].imshow(mask)
ax[0].set_title(
    "Ground Truth"
)

ax[1].imshow(prediction)
ax[1].set_title(
    "Prediction"
)

plt.show()
```
