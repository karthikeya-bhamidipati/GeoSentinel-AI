```python
import torch
import torch.nn as nn
import torch.optim as optim
```

```python
x = torch.tensor([
    [1.0],
    [2.0],
    [3.0],
    [4.0]
])

y = torch.tensor([
    [2.0],
    [4.0],
    [6.0],
    [8.0]
])
```

```python
model = nn.Linear(
    in_features=1,
    out_features=1
)

print(model)
```

```python
with torch.no_grad():

    prediction = model(x)

print(prediction)
```

```python
criterion = nn.MSELoss()

print(criterion)
```

```python
optimizer = optim.SGD(
    model.parameters(),
    lr=0.01
)

print(optimizer)
```

```python
prediction = model(x)

loss = criterion(
    prediction,
    y
)

print(
    "Loss:",
    loss.item()
)
```

```python
optimizer.zero_grad()

loss.backward()

optimizer.step()
```

```python
prediction = model(x)

new_loss = criterion(
    prediction,
    y
)

print(
    "New Loss:",
    new_loss.item()
)
```

```python
epochs = 100

loss_history = []

for epoch in range(epochs):

    prediction = model(x)

    loss = criterion(
        prediction,
        y
    )

    optimizer.zero_grad()

    loss.backward()

    optimizer.step()

    loss_history.append(
        loss.item()
    )

    if epoch % 10 == 0:

        print(
            f"Epoch {epoch}: "
            f"{loss.item():.4f}"
        )
```

```python
with torch.no_grad():

    prediction = model(x)

print(prediction)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.plot(loss_history)

plt.title(
    "Training Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.show()
```

```python
learning_rates = [
    0.1,
    0.01,
    0.001
]

for lr in learning_rates:

    print(
        f"Learning Rate: {lr}"
    )
```
